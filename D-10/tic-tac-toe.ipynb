```python
import math

b = [" " for _ in range(9)]

def p():
    print(b)
    print()

def w(x):
    c = [[0,1,2],[3,4,5],[6,7,8],
         [0,3,6],[1,4,7],[2,5,8],
         [0,4,8],[2,4,6]]
    for i in c:
        if x[i[0]] == x[i[1]] == x[i[2]] != " ":
            return x[i[0]]
    if " " not in x:
        return "D"
    return None

def m(x, t):
    r = w(x)
    if r == "O":
        return 1
    if r == "X":
        return -1
    if r == "D":
        return 0
    
    if t:
        s = -math.inf
        for i in range(9):
            if x[i] == " ":
                x[i] = "O"
                s = max(s, m(x, False))
                x[i] = " "
        return s
    else:
        s = math.inf
        for i in range(9):
            if x[i] == " ":
                x[i] = "X"
                s = min(s, m(x, True))
                x[i] = " "
        return s

def best():
    s = -math.inf
    move = 0
    for i in range(9):
        if b[i] == " ":
            b[i] = "O"
            v = m(b, False)
            b[i] = " "
            if v > s:
                s = v
                move = i
    return move

def play():
    while True:
        p()
        n = int(input()) - 1
        if b[n] == " ":
            b[n] = "X"
        if w(b):
            break
        k = best()
        b[k] = "O"
        if w(b):
            break
    p()
    print(w(b))

play()
```
