```python
# A – B
# A – C
# B – C
# B – D
# C – D
# C – E
# D – E
# D – F
# E – F
# E – G
# F – G

g={'A':['B','C'],
   'B':['A','C','D'],
   'C':['A','B','D','E'],
   'D':['B','C','E','F'],
   'E':['C','D','F','G'],
   'F':['D','E','G'],
   'G':['E','F']}

v=list(g)
n=len(v)

def ok(x,c,a):
    for y in g[x]:
        if a.get(y)==c:
            return False
    return True

def bt(m,a,i):
    if i==n:
        return True
    x=v[i]
    for c in range(1,m+1):
        if ok(x,c,a):
            a[x]=c
            if bt(m,a,i+1):
                return True
            del a[x]
    return False 

for m in range(1,n+1):
    a={}
    if bt(m,a,0):
        print("coloring possible with",m,"colors")
        for x in v:
            print("vertex",x,"→ color",a[x])
        print("minimum number of colors required =",m)
        break
```
