```python
from queue import Queue

def is_safe(M, D):
    return M == 0 or M >= D

def mds():
    q = Queue()
    visited = set()
    parent = {}

    start = (3, 3, 0)
    q.put(start)
    visited.add(start)
    parent[start] = None

    moves = [(1,0), (2,0), (0,1), (0,2), (1,1)]

    while not q.empty():
        M, D, B = q.get()

        if (M, D, B) == (0, 0, 1):
            printsol(parent, (M, D, B))
            return

        for m, d in moves:
            if B == 0:  
                nM = M - m
                nD = D - d
                nB = 1
            else:       
                nM = M + m
                nD = D + d
                nB = 0

            if 0 <= nM <= 3 and 0 <= nD <= 3:
                rightM = 3 - nM
                rightD = 3 - nD

                if is_safe(nM, nD) and is_safe(rightM, rightD):
                    state = (nM, nD, nB)
                    if state not in visited:
                        visited.add(state)
                        parent[state] = (M, D, B)
                        q.put(state)

    print("No solution possible")


def printsol(parent, goal):
    path = []
    while goal is not None:
        path.append(goal)
        goal = parent[goal]

    path.reverse()
    for state in path:
        print(state)
        
mds()
```
