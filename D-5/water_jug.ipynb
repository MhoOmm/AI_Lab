```python
from queue import Queue

def water_jug(A, B, C):
    q = Queue()          
    visited = set()
    parent = {}

    start = (0, 0)
    q.put(start)        
    visited.add(start)
    parent[start] = None

    while not q.empty():
        x, y = q.get()   

        
        if x == C or y == C:
            printsol(parent, (x, y))
            return

        
        nxt = [
            (A, y),                          # fill  A
            (x, B),                          # fill  B
            (0, y),                          # empty  A
            (x, 0),                          # empty  B
            (x - min(x,B-y), y + min(x,B-y)),  #  A → B
            (x + min(y,A-x), y - min(y,A-x))   #  B → A
        ]

        for state in nxt:
            if state not in visited:
                visited.add(state)
                parent[state] = (x, y)
                q.put(state)     

    print("no solution")
    
    
def printsol(parent, goal):
    path = []
    while goal is not None:
        path.append(goal)
        goal = parent[goal]

    path.reverse()

    print("solution path:")
    for state in path:
        print(state)
        
     
     
# Main function    
A = 4
B = 3
C = 2
water_jug(A,B,C)
```
