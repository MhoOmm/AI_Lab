```python
import random

n = 4
g = [["" for _ in range(n)] for _ in range(n)]

# place agent
ax, ay = 0, 0

# random positions
wx, wy = random.randint(0,3), random.randint(0,3)
gx, gy = random.randint(0,3), random.randint(0,3)

p = []
for i in range(2):
    x, y = random.randint(0,3), random.randint(0,3)
    p.append((x,y))

while (wx,wy)==(0,0): wx,wy=random.randint(0,3),random.randint(0,3)
while (gx,gy)==(0,0): gx,gy=random.randint(0,3),random.randint(0,3)

def near(x,y,a,b):
    return abs(x-a)+abs(y-b)==1

while True:

    print("\nAgent Position:",(ax+1,ay+1))

    b = any(near(ax,ay,x,y) for x,y in p)
    s = near(ax,ay,wx,wy)
    gl = (ax,ay)==(gx,gy)

    print("Percepts:")
    if b: print("Breeze detected!")
    if s: print("Stench detected!")
    if gl: print("Glitter! Gold here")

    if (ax,ay)==(wx,wy):
        print("Wumpus ate the agent")
        break

    if (ax,ay) in p:
        print("Agent fell into pit")
        break

    if gl:
        print("Agent found gold and wins")
        break

    print("\nMove Options:")
    if ax>0: print("1. Up")
    if ax<3: print("2. Down")
    if ay>0: print("3. Left")
    if ay<3: print("4. Right")

    c = int(input("Move: "))

    if c==1 and ax>0: ax-=1
    elif c==2 and ax<3: ax+=1
    elif c==3 and ay>0: ay-=1
    elif c==4 and ay<3: ay+=1
```
