```python
from queue import PriorityQueue

directions = [(0,1),(0,-1),(1,0),(-1,0)]

goal = [
    [1,2,3],
    [4,5,6],
    [7,8,0]
]

def h1(s):
    c = 0
    for i in range(3):
        for j in range(3):
            if s[i][j] != 0 and s[i][j] != goal[i][j]:
                c = c + 1
    return c

def h2(s):
    d = 0
    for i in range(3):
        for j in range(3):
            v = s[i][j]
            if v != 0:
                for x in range(3):
                    for y in range(3):
                        if goal[x][y] == v:
                            d = d + abs(i-x) + abs(j-y)
    return d

def find0(s):
    for i in range(3):
        for j in range(3):
            if s[i][j] == 0:
                return i,j

def copy(s):
    n = []
    for i in range(3):
        r = []
        for j in range(3):
            r.append(s[i][j])
        n.append(r)
    return n

def to_tuple(s):
    t = []
    for i in range(3):
        for j in range(3):
            t.append(s[i][j])
    return tuple(t)

def astar(start,h):
    q = PriorityQueue()
    q.put((0,start))
    
    g = {}
    g[to_tuple(start)] = 0
    
    p = {}
    p[to_tuple(start)] = None
    
    v = set()
    
    n = 0
    
    while not q.empty():
        f,s = q.get()
        n = n + 1
        
        if s == goal:
            path = []
            k = to_tuple(s)
            while k is not None:
                path.append(k)
                k = p[k]
            path.reverse()
            return path,n
        
        v.add(to_tuple(s))
        x,y = find0(s)
        
        for dx,dy in directions:
            nx = x + dx
            ny = y + dy
            
            if 0 <= nx < 3 and 0 <= ny < 3:
                ns = copy(s)
                ns[x][y] = ns[nx][ny]
                ns[nx][ny] = 0
                t = to_tuple(ns)
                
                if t not in v:
                    ng = g[to_tuple(s)] + 1
                    if t not in g or ng < g[t]:
                        g[t] = ng
                        q.put((ng + h(ns),ns))
                        p[t] = to_tuple(s)
    
    return None,n

start = [
    [1,2,3],
    [4,0,5],
    [6,7,8]
]

p1,n1 = astar(start,h1)
p2,n2 = astar(start,h2)

print("H1 path length")
print(len(p1)-1)
print("H1 nodes explored")
print(n1)

print("H2 path length")
print(len(p2)-1)
print("H2 nodes explored")
print(n2)
```
