```python
from queue import Queue

def bfs(start, adj, visited):
    q = Queue()
    q.put(start)
    visited[start] = True

    print("BFS Traversal:")
    while not q.empty():
        top = q.get()
        print(top, end=" ")

        for nei in adj[top]:
            if not visited[nei]:
                visited[nei] = True
                q.put(nei)
    return visited


def dfs(node, adj, visited):
    visited[node] = True
    print(node, end=" ")

    for nei in adj[node]:
        if not visited[nei]:
            dfs(nei, adj, visited)


def search_node(num, visited, n):
    if num < 0 or num >= n:
        print("Surpassed index value")
        return

    if visited[num]:
        print("Element found")
    else:
        print("Element not found")


n = 5
adj = [[] for _ in range(n)]

edges = [(0,1),(0,2),(1,2),(1,3),(2,4)]
for u, v in edges:
    adj[u].append(v)


choice = input("Enter traversal (bfs/dfs): ").lower()
visited = [False] * n

start = int(input("Enter the starting node"))

if choice == 'bfs':
    bfs(start,adj,visited)
else:
    dfs(start,adj,visited)

num = int(input("Enter node to be searched: "))
search_node(num, visited, n)
```
