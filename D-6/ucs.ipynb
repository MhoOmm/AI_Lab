```python
from queue import PriorityQueue

n = 5
adj = [[] for _ in range(n)]
INT_MAX = 2**31 - 1

edges = [(0,1,1),(0,2,3),(1,2,4),(1,3,10),(2,4,8),(1,4,7)]
for u, v, wt in edges:
    adj[u].append([v, wt])
    
print("Adjacency list:", adj)

start = int(input("Enter the element to start: "))
dist = [INT_MAX] * n
dist[start] = 0

pq = PriorityQueue()
pq.put((0, start))

while not pq.empty():
    currwt, node = pq.get()
    
    if currwt > dist[node]: 
        continue
    
    for neig, neigwt in adj[node]:
        newwt = currwt + neigwt
        if newwt < dist[neig]:
            dist[neig] = newwt
            pq.put((newwt, neig))
            
target = int(input("Enter the destination node: "))
print("shortest path from node", start, "to node", target, "is:", dist[target])
```
